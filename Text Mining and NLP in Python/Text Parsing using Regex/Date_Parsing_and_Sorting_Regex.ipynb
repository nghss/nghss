# Assignment 1

In this assignment, you'll be working with messy medical data and using regex to extract relevant infromation from the data. 

Each line of the `dates.txt` file corresponds to a medical note. Each note has a date that needs to be extracted, but each date is encoded in one of many formats.

The goal of this assignment is to correctly identify all of the different date variants encoded in this dataset and to properly normalize and sort the dates. 

Here is a list of some of the variants you might encounter in this dataset:
* 04/20/2009; 04/20/09; 4/20/09; 4/3/09
* Mar-20-2009; Mar 20, 2009; March 20, 2009;  Mar. 20, 2009; Mar 20 2009;
* 20 Mar 2009; 20 March 2009; 20 Mar. 2009; 20 March, 2009
* Mar 20th, 2009; Mar 21st, 2009; Mar 22nd, 2009
* Feb 2009; Sep 2009; Oct 2010
* 6/2008; 12/2009
* 2009; 2010

Once you have extracted these date patterns from the text, the next step is to sort them in ascending chronological order accoring to the following rules:
* Assume all dates in xx/xx/xx format are mm/dd/yy
* Assume all dates where year is encoded in only two digits are years from the 1900's (e.g. 1/5/89 is January 5th, 1989)
* If the day is missing (e.g. 9/2009), assume it is the first day of the month (e.g. September 1, 2009).
* If the month is missing (e.g. 2010), assume it is the first of January of that year (e.g. January 1, 2010).
* Watch out for potential typos as this is a raw, real-life derived dataset.

With these rules in mind, find the correct date in each note and return a pandas Series in chronological order of the original Series' indices. **This Series should be sorted by a tie-break sort in the format of ("extracted date", "original row number").**

For example if the original series was this:

    0    1999
    1    2010
    2    1978
    3    2015
    4    1985

Your function should return this:

    0    2
    1    4
    2    0
    3    1
    4    3

Your score will be calculated using [Kendall's tau](https://en.wikipedia.org/wiki/Kendall_rank_correlation_coefficient), a correlation measure for ordinal data.

*This function should return a Series of length 500 and dtype int.*

```python
# begin processing for each format type  
# type 1: change /yy or -yy to /19yy or -19yy 
import pandas as pd 
import re

# define a function to apply the changes 
def fix_two_digit_year(date): 
    if pd.notna(date): # check value not NaN 
        match = re.search(r'[-/]{1}\d{2}$', date) # check if ending is /YY or -YY
        if match: 
            return f"{date[:-2]}19{date[-2:]}"  # add 19 before YY 
    return date  # return unchanged if already YYYY 
    
dates['format1'] = dates['format1'].apply(fix_two_digit_year)
dates['format1']
```

```python
# type 2: change Month name to numeric month 
# then reformat date to get mm/dd/yyyy
from datetime import datetime 

def convert_Month_dd_yyyy(date): 
    if pd.notna(date) and isinstance(date,str): # ensure non null and string 
        try: 
            dt = datetime.strptime(date, "%B %d, %Y") # full month name 
        except ValueError: 
            try: 
                dt = datetime.strptime(date, "%b %d, %Y") # abbreviated month name 
            except ValueError: 
                return np.nan # return NaN if parsing fails 
        return dt.strftime("%m/%d/%Y")
    return date # return NaN if unchanged 

dates['format2'] = dates['format2'].apply(convert_Month_dd_yyyy)
dates['format2'].unique()
```

```python
# type 3: dd Month yyyy -- change Month name to numeric month 
# then reformat date to get mm/dd/yyyy
from datetime import datetime 
import numpy as np 

def convert_ddmmyyyy(date): 
    if pd.notna(date) and isinstance(date,str): # ensure non null and string
        try: 
            dt = datetime.strptime(date, "%d %B %Y")  # full month name 
        except ValueError: 
            try: 
                dt = datetime.strptime(date, "%d %b %Y") # abbreviated month name 
            except ValueError: 
                return np.nan # return NaN if parsing fails 
            
        return dt.strftime("%m/%d/%Y")
    return date # return unchanged if NaN 

dates['format3'] = dates['format3'].apply(convert_ddmmyyyy) 
dates['format3'].unique()
    
```

```python
# type 4: Month yyyy -- change to mm/01/yyyy
from datetime import datetime 
import numpy as np

def convert_mm_01_yyyy(date): 
    if pd.notna(date) and isinstance(date,str): # ensure it is non-null string 
        try: 
            # parse using known formats for the date 
            dt = datetime.strptime(date, "%B %Y")  # full month name 
        except ValueError: 
            try: 
                dt = datetime.strptime(date, "%b %Y")  # abbreviated month name 
            except ValueError: 
                try:
                    dt = datetime.strptime(date, "%b, %Y")
                except ValueError: 
                    try: 
                        dt = datetime.strptime(date, "%B, %Y")
                    except ValueError: 
                        return np.nan # return NaN if parsing fails 
            
        return dt.strftime("%m/01/%Y") 
    
    return date # return unchanged if NaN or invalid

dates['format4'] = dates['format4'].apply(convert_mm_01_yyyy)
dates['format4'].unique()
```

```python
# type 5: mm/yyyy -- change to mm/01/yyyy

# define a function to apply the changes 
def fix_missing_dd(date): 
    if pd.notna(date): # check value not NaN 
        match = re.search(r'\d{1,2}[-/]{1}\d{4}$', date) # check if ending is MM/YYYY or MM-YYYY
        if match: 
            return f"{date[:-4]}01/{date[-4:]}"  # add 01/ before YYYY 
    return date  
    
dates['format5'] = dates['format5'].apply(fix_missing_dd)
dates['format5'].unique()
```

```python
# type 6: yyyy -- strip off unwanted symbols in front and behind 
# change to 01/01/yyyy 

# define a function to apply the changes 
def fix_missing_dd_mm(date): 
    if pd.notna(date): # check value not NaN 
        match = re.search(r'\d{4}', date) # check if ending is YYYY only
        if match: 
            year = match.group(0) # get matched year 
            return f"01/01/{year}"  # add 01/01/ before YYYY 
    return date  
    
dates['format6'] = dates['format6'].apply(fix_missing_dd_mm)
dates['format6'].unique()
```

```python
# merge dates to form a new column that contains all processed dates 
dates['extracted_date'] = dates.bfill(axis=1).iloc[:,0]
dates['extracted_date'].unique()
```
